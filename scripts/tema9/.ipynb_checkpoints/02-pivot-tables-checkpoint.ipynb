# Tablas de Pivotaje

````python
import numpy as np
import pandas as pd
import seaborn as sns
````

````python
titanic = sns.load_dataset("titanic")
titanic.head()
````

````python
titanic.groupby("sex")[["survived"]].mean()
````

````python
titanic.groupby(["sex", "pclass"])["survived"].aggregate("mean").unstack()
````

````python
titanic.pivot_table("survived", index="sex", columns="pclass") # forma más fácil de entender
````

````python
age = pd.cut(titanic["age"], bins=[0, 18, 25, 35, 50, 65, 100],
             include_lowest=True, precision=1, right=False)
age
````

````python
titanic.pivot_table("survived", index=["sex", age], columns="class")
````

````python
fare = pd.qcut(titanic["fare"], q=4, labels=["Q1", "Q2", "Q3", "Q4"])
fare
````

````python
titanic.pivot_table("survived", index=["sex", age], columns=[fare, "pclass"])
````

````python
titanic.pivot_table(index="sex", columns="class",
                   aggfunc={"survived": sum, "fare": "mean"})
````

````python
titanic.pivot_table("survived", index="sex", columns="class", margins=True, margins_name="Total")
````
