# Análisis de nacimiento de bebés en US

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
births = pd.read_csv("../../data/us-births.csv")
```

```python
births.shape
```

```python
births.head()
```

```python
births.tail()
```

```python
births.pivot_table("births", index="year", columns="gender", aggfunc="sum")
```

```python
births["decade"] = 10*(births["year"]//10)
```

```python
births.head()
```

```python
births.pivot_table("births", index="decade", columns="gender", aggfunc="sum")
```

```python
sns.set()
```

```python
plt.figure(figsize=(10, 8))
births.pivot_table("births", index="year", columns="gender", aggfunc="sum").plot()
plt.ylabel("Total de Nacimientos por Año")
plt.show()
```

```python
quartiles = np.percentile(births["births"], [25, 50, 75])
quartiles
```

```python
med = quartiles[1]
IQR = quartiles[2] - quartiles[0]
sig = 0.75*IQR
```

```python
births = births.query("(births > @med - 5*@sig) & (births < @med + 5*@sig)")
```

```python
births.shape
```

```python
births["day"] = births["day"].astype(int)
```

```python
births.head()
```

```python
births.index = pd.to_datetime(10000*births.year + 100*births.month + births.day, format="%Y%m%d")
```

```python
births.head()
```

```python
births["dayofweek"] = births.index.dayofweek
```

```python
births.head()
```

```python
births.pivot_table("births", index="dayofweek", columns="decade", aggfunc="mean").plot()
plt.gca().set_xticklabels(["", "Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo"])
plt.ylabel("Media de nacimientos por día")
plt.show()
```

```python
births_by_date = births.pivot_table("births", index=[births.index.month, births.index.day], aggfunc="mean")
```

```python
births_by_date.head()
```

```python
births_by_date.index = [pd.datetime(2020, month, day) for (month, day) in births_by_date.index]
```

```python
births_by_date.head()
```

```python
plt.figure(figsize=(12, 4))
births_by_date.plot()
plt.show()
```
