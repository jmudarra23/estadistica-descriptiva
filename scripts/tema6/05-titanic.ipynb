## Tablas de contingencia de dos variables

```python
survived_sex = pd.crosstab(index=titanic["survived"],
                          columns=titanic["sex"])
survived_sex.index = ["died", "survived"]
survived_sex
```

```python
survived_class = pd.crosstab(index=titanic["survived"], 
                            columns=titanic["pclass"],
                            margins=True)

survived_class.index = ["murio", "sobrevivio", "total_clase"]
survived_class.columns = ["primera", "segunda", "tercera", "total_superv"]
survived_class
```

### Frecuencias relativas globales

```python
survived_class/survived_class.loc["total_clase", "total_superv"] # porcentajes en base al total de supervivientes
```

### Frecuencias relativas marginales

```python
survived_class/survived_class.loc["total_clase"] # porcentaje en base a los supervivientes de cada una de las clases
```

```python
survived_class.div(survived_class.loc["total_clase"], axis=1) # otra forma porcentaje en base a los supervivientes de cada una de las clases
```

```python
survived_class.T/survived_class["total_superv"] # frecuencias relativas marginales por filas
```

```python
survived_class.div(survived_class["total_superv"], axis=0) # otra forma frecuencias relativas marginales por filas
```
