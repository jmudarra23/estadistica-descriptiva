# Operadores de decisión

```
si condición booleana:
    código que se ejecute si la condición es verdad
    ...
    ...
    ...
siguiente código que se ejecuta siempre
```

````python
x = int(input("Escribe un número:"))
````

````python
if x == 5:
    print("Has escrito el número 5")
````

````python
x = int(input("Escribe un número:"))

if x < 5:
    print("Has escrito un número menor que 5")
elif x < 10:
    print("Has escrito un número menor que 10 y menor o igual a 5")
else:
    print("El número que has escrito es mayor o igual a 10")
````

````python
x = int(input("Escribe un número:"))

if x > 0 and x < 10:
    print("Has escrito un número en el rango [0, 10]")
````

````python
x = int(input("Escribe un número:"))

if x < 0 or x > 10:
    print("Tu número escrito no se encuentra dentro del rango [0, 10]")
````
