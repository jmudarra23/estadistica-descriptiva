# Funciones matemáticas

```python
math.exp(3) # es lo mismo que e^3, es más preciso que hacer math.e**3
```

```python
math.e**3 # en los últimos dígitos ya falla
```

```python
math.pow(math.e, 3)
```

```python
math.expm1(1) # exponencial - 1, precisión más correcta
```

```python
math.exp(1) - 1 # perdemos precisión
```

```python
math.exp(1e-05) - 1
```

```python
math.expm1(1e-05) # precisión más correcta
```
