# Copias y vistas de array

```python
x = np.arange(10)
```

```python
y = x # así se crea una referencia, no una copia
```

```python
y.shape = (2, 5)
y
```

```python
x
```

```python
z = x.copy()
z
```

```python
z is x # z es una copia de x
```

```python
y is x
```

```python
t = x.view()
t
```

```python
t.shape = (5, 2)
t
```

```python
x
```

```python
s = x[0:2, 1:3] # esto es una vista
s
```

```python
s[:] = 5
s
```

```python
x
```

```python
y
```

```python
z
```
