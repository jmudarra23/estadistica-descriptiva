# Listas

Son contenedores que permiten almacenar una colección de objetos en un orden determinado. Se pueden acceder, añadir o eliminar objetos de forma sencilla.

```python
L1 = [] # Lista vacía
```

```python
L2 = ["a", "e", "i", "o", "u"] # Lista con valores iniciales
```

```python
L3 = [2, 3, 5, 7, 11] # Lista con valores enteros
```

```python
L4 = ["a", 5, True, "Jesús Mudarra"] # Lista hetereogénea con varios tipos de datos diferentes
```

```python
L5 = [0]*10
L5
```

```python
L4[0] # Las listas en Python empiezan en la posición número 0
```

```python
L4[1]
```

```python
L5[4] = 3.1415 # Modificación de valores muy cómoda
L5
```

```python
L5[13] = 8 # Error índice fuera rango
```

```python
L5[13] # No existe el índice 13
```

```python
L4[0:2] # Accesos de posiciones múltiples (incluye la primera, pero no la última)
```

```python
L5[1:4] = [1, 2, 3]
L5
```

```python
L5.append(8) # Añadir un nuevo elemento a la lista
L5
```

```python
L5.remove(3) # Elimina la primera ocurrencia del valor de la lista
```

# Bucles

Los bucles **for** iteran sobre una colección (una lista, un string, un objeto iterable).

```python
for numero in L:
    print(numero)
```

```python
numero
```

```python
for x in range(0, 10): # range(10) = range(0, 10)
    print(x)
```

```python
for c in "Jesús Mudarra":
    print(c)
```

```python
notas = [3.5, 6.7, 8, 9, 4.5]
suma = 0
for nota in notas:
    suma += nota
print(suma/len(notas))
```

El bucle **while** se repite hasta que la condición booleana deja de ser cierta.

```python
count = 0
while count < 10:
    print(count)
    count += 1
```

```python
primos = [1, 3, 5, 7, 11, 13, 17, 19, 23, 29]
for idx, p in enumerate(primos):
    print(idx, p)
```

```python
for idx in range(len(primos)):
    print(idx, primos[idx])
```

### La criba de Eratóstenes (a mi manera)

```python
nums = list(range(2, 121)) # números naturales entre 2 y 120
i = 2
while i**2 < 120:
    for num in nums:
        if num <= i:
            continue
        elif num % i == 0:
            nums.remove(num)
        else:
            pass
    i += 1

print("Los números primos comprendidos entre 2 y 120 son:", nums)
```

## Ejercicio propuesto

```python
num_collection_list = str(input("Escribe una colección de números:")).split(",")
```

```python
for i in range(len(num_collection_list)):
    num_collection_list[i] = int(num_collection_list[i])
```

```python
num_collection_tuple = tuple(num_collection_list)
```

```python
suma = 0
for n in num_collection_tuple:
    suma += n
    
print(suma/len(num_collection_tuple))
```
