# Diagrama de Caja y Bigotes

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(19880519)
```

```python
dist1 = 100*np.random.rand(50)
dist2 = 50*np.ones(25)
dist3 = 100*np.random.rand(10) + 100
dist4 = -100*np.random.rand(10)
data = np.concatenate((dist1, dist2, dist3, dist4))
```

```python
fig1, ax1 = plt.subplots()
ax1.set_title("Boxplot básico")
ax1.boxplot(data)
plt.show()
```

```python
fig2, ax2 = plt.subplots()
ax2.set_title("Boxplot muescas")
ax2.boxplot(data, notch=True)
plt.show()
```

```python
green_diamond = dict(markerfacecolor="g", marker="D")
fig3, ax3 = plt.subplots()
ax3.set_title("Boxplot con outliers personalizados")
ax3.boxplot(data, flierprops=green_diamond)
plt.show()
```

```python
fig4, ax4 = plt.subplots()
ax4.set_title("Boxplot sin outliers")
ax4.boxplot(data, showfliers=False) # sin que aparezcan los outliers
plt.show()
```

```python
red_square = dict(markerfacecolor="r", marker="s")
fig5, ax5 = plt.subplots()
ax5.set_title("Boxplot personalizado y en horizontal")
ax5.boxplot(data, vert=False, flierprops=red_square,
            whis=0.75) # haciendo más pequeños los bigotes (saldrán más outliers)
plt.show()
```

```python
dist1 = 100*np.random.rand(50)
dist2 = 40*np.ones(25)
dist3 = 100*np.random.rand(10) + 100
dist4 = -100*np.random.rand(10)
data2 = np.concatenate((dist1, dist2, dist3, dist4))
```

```python
data.shape = (-1, 1)
data2.shape = (-1, 1)
```

```python
fulldata = np.array([data, data2, data2[::2, 0]], dtype="object")
```

```python
fig6, ax6 = plt.subplots()
ax6.set_title("Múltiples muestras de diferente naturaleza")
ax6.boxplot(fulldata)
plt.show()
```
