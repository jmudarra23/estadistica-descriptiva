## Pregunta 1

Crea una función que reciba los tres coeficientes $a$, $b$ y $c$ para resolver una ecuación de segundo grado. Muestra la solución por pantalla y ayúdate de la librería `math` para acceder a la función raíz cuadrada.

```python
import math

def ec2grado(a, b, c):
    discriminante = b**2 - 4*a*c
    if discriminante > 0:
        x_1 = (-b + math.sqrt(discriminante))/(2*a)
        x_2 = (-b - math.sqrt(discriminante))/(2*a)
        print("Ya que el discriminante es mayor que cero entonces la ecuación de segundo grado tiene dos raíces reales x1 = {} y x2 = {}".format(x_1, x_2))
    elif discriminante == 0:
        x = -b/(2*a)
        print("Ya que el discriminante es cero entonces la ecuación de segundo grado tiene sólo una raíz real x = {}".format(x))
    else:
        print("Ya que el discriminante es menor que cero entonces la ecuación no tiene resolución real.")
```

```python
ec2grado(4, 4, 1)
```

## Pregunta 2

Crea una función que lea una frase de teclado y nos diga si es o no un palíndromo (frase que se lee igual de izquierda a derecha o al revés como por ejemplo “La ruta nos aportó otro paso natural”).

```python
def string_cleaning(s):
    vocales_con_tilde = {"á": "a", "é": "e", "í": "i", "ó": "o", "ú": "u"}
    for key, value in vocales_con_tilde.items():
        s = s.replace(key, value) # reemplazo de vocales
        s = s.replace(" ", "") # quitar espacios
    return s

def is_palindrom(s):
    for idx, c in enumerate(s):
        if s[idx] == s[len(s) -1 -idx]:
            continue
        else:
            print("La frase NO es palíndroma")
            return # salir de la función
    print("La frase es palíndroma")
```

```python
frase = str(input("Escribe aquí tu frase:")).lower()
frase = string_cleaning(frase)
```

```python
is_palindrom(frase)
```

## Pregunta 6

Crea una función que dado un número $N$ nos diga si es primo o no (tiene que ir dividiendo por todos los números x comprendidos entre $2$ y el número $N − 1$ y ver si la división de $\frac{N}{x}$ tiene resto cero o no).

```python
def is_prime(N):
    for x in range(2, N):
        if N % x == 0:
            print("El número NO es primo")
            return
    print("El número es primo")
```

```python
n = 11

is_prime(n)
```

## Pregunta 8

Crea una función que calcule el máximo común divisor de dos números introducidos por el usuario por teclado.

```python
n1 = int(input("Introduzca el primer número:"))
n2 = int(input("Introduzca el segundo número:"))

multiplos_1 = []
multiplos_2 = []

# calcular múltiplos del primer número
for mult in range(1, n1 + 1):
    if n1 % mult == 0:
        multiplos_1.append(mult)

# calcular múltiplos del segundo número
for mult in range(1, n2 + 1):
    if n2 % mult == 0:
        multiplos_2.append(mult)

# quedarnos con último múltiplo común (el más alto) de ambas listas
for i in multiplos_1:
    if i in multiplos_2:
        mcd = i

print("El máximo común divisor de {} y {} es {}".format(n1, n2, mcd))
```

## Pregunta 9

Investiga el Cifrado del César y crea una función que lo reproduzca en Python. Cada letra del mensaje original se desplaza tres posiciones en el alfabeto estándar. La A se convierte en la D, la B se convierte en la E, la C se convierte en la F... y cuando se acaba el alfabeto se le vuelve a dar la vuelta: la X se convierte en la A, la Y en la B y la Z en la C. Los números no sufren ninguna modificación.

```python
frase = str(input("Introduzca una frase:")).upper()
```

```python
def cifrado_cesar(n, s):
    cesar_list = []
    
    # si el caracter leído es un espacio guardamos un espacio en la lista cesar_list
    for c in s:
        if c == " ":
            cesar_list.append(" ")
            continue
            
        c_new = chr(ord(c) + n)
        if ord(c_new) > 90:
            c_new = chr(ord(c_new) - 26)
        cesar_list.append(c_new)
        
    frase_cifrada = "".join(cesar_list)
    return frase_cifrada
```

```python
cifrado_cesar(3, frase)
```
