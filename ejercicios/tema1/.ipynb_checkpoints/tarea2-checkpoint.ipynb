## Pregunta 1

Escribe una secuencia de instrucciones que permitan leer un número real por pantalla y que muestre si el número es positivo o no.

```python
num = float(input("Introduce un número real:"))

if num >= 0:
    print("El número {} es positivo".format(num))
else:
    print("El número {} es negativo".format(num))
```

## Pregunta 2

Escribe una secuencia de instrucciones que permitan leer un número real por pantalla y que muestre si el número está en el rango entre −5 y 5, ambos incluidos.

```python
num = float(input("Introduce un número real:"))

if num >= -5 and num <= 5:
    print("El número {} se encuentra en el rango (-5, 5)".format(num))
else:
    print("El número {} se encuentra FUERA del rango (-5, 5)".format(num))
```

## Pregunta 3

Escribe una secuencia de instrucciones que permitan leer las coordenadas de un punto $(x, y)$ e indique en cuál de los cuatro cuadrantes se encuentra dicho punto.

Si $x = 0$, deberás indicar que el punto se encuentra sobre el eje vertical.

Si $y = 0$, deberás indicar que el punto se encuentra sobre el eje horizontal.

Si tanto $x = 0$ como $y = 0$, entonces deberás indicar que el punto se trata del origen de coordenadas.

```python
x = int(input("Introduce la coordenada x:"))
y = int(input("Introduce la coordenada y:"))
```

```python
if x == 0 and y == 0:
    print("El punto ({}, {}) se encuentra en el origen de coordenadas".format(x, y))
elif x > 0 and y > 0:
    print("El punto ({}, {}) se encuentra en el primer cuadrante".format(x, y))
elif x < 0 and y > 0:
    print("El punto ({}, {}) se encuentra en el segundo cuadrante".format(x, y))
elif x < 0 and y < 0:
    print("El punto ({}, {}) se encuentra en el tercer cuadrante".format(x, y))
elif x > 0 and y < 0:
    print("El punto ({}, {}) se encuentra en el cuarto cuadrante".format(x, y))
elif x == 0:
    print("El punto ({}, {}) se encuentra sobre el eje vertical".format(x, y))
elif y == 0:
    print("El punto ({}, {}) se encuentra sobre el eje horizontal".format(x, y))
```

## Pregunta 4

Escribe una secuencia de instrucciones que permitan leer dos números enteros y muestre el cociente de la división entera y el resto de la división entera.


```python
num1 = int(input("Introduzca el dividendo de la división:"))
num2 = int(input("Introduzca el divisor de la división:"))
```

```python
cociente = num1//num2
resto = num1 % num2
```

```python
print("El cociente de la división {}/{} es igual a {}".format(num1, num2, cociente))
print("El resto de la división {}/{} es igual a {}".format(num1, num2, resto))
```

## Pregunta 5

Escribe una secuencia de instrucciones que permitan leer un número entero y determinar si es cuadrado perfecto o no (piensa la mejor forma de hacerlo con lo que has aprendido hasta ahora).

```python
num = int(input("Introduzca el número:"))
```

```python
import math
if math.sqrt(num).is_integer():
    print("El número {} es un cuadrado perfecto".format(num))
else:
    print("El número {} NO es un cuadrado perfecto".format(num))
```

## Pregunta 6

Escribe una expresión que permita determinar si un número entero positivo puede corresponder a un año bisiesto o no. Se consideran años bisiestos aquellos cuyo número es divisible por cuatro excepto los años que son múltiplos de 100, a no ser que lo sean de 400 (por ejemplo el año 2000 fue bisiesto pero el 2100 no lo será).

```python
year = int(input("Introduce un año:"))
```

```python
if year % 4 == 0 and not year % 100 == 0 or year % 400 == 0:
    print("El año es bisiesto")
else:
    print("El año NO es bisiesto")
```

# Pregunta 7

Busca la imagen de un tablero de ajedrez en Google y fíjate en la nomenclatura de las casillas. Escribe una secuencia que lea una letra y un número de teclado correspondiente a una casilla de un tablero de ajedrez y que indique si esta casilla es negra o blanca.

```python
from IPython.display import Image
from IPython.core.display import HTML 
Image(url="https://www.ajedrez21.com/17107-superlarge_default/tablero-plastico-competicion.jpg", width=300, height=300)
```

```python
numeros = [1, 2, 3, 4, 5, 6, 7, 8]
letras = ["a", "b", "c", "d", "e", "f", "g", "h"]
```

```python
l = input("Introduzca una letra:").lower()
n = int(input("Introduce un número:"))

pos_numero = numeros.index(n)
pos_letra = letras.index(l)

if pos_letra % 2 == 0 and pos_numero % 2 == 0:
    print("Casilla Negra")
elif pos_letra % 2 != 0 and pos_numero % 2 != 0:
    print("Casilla Negra")
else:
    print("Casilla Blanca")
```

```python
# Consultar colores de todas las casillas

for letra in letras:
    print("\nColumna letra {}:\n----------------".format(letra.upper()))
    for numero in numeros:
        if letras.index(letra) % 2 == 0 and numeros.index(numero) % 2 == 0:
            print("Casilla {}{} Negra".format(letra, numero))
        elif letras.index(letra) % 2 != 0 and numeros.index(numero) % 2 != 0:
            print("Casilla {}{} Negra".format(letra, numero))
        else:
            print("Casilla {}{} Blanca".format(letra, numero))
```
