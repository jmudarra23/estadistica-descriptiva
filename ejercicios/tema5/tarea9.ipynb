```python
df = pd.read_csv("../../data/run.csv")
df["variacion_pulso"] = (df["pulso.despues"] - df["pulso.antes"])/df["pulso.antes"]
```

## Pregunta 3

Calcula el porcentaje medio de variación del pulso por minuto entre antes y después de hacer ejercicio y compara el valor de los que hacen ejercicio habitualmente y los que no. ¿Observas mucha diferencia?

```python
df.groupby("hace.deporte")["variacion_pulso"].mean()
```

Existe mucha diferencia. Las personas que hacen deporte tienen una media de 26.22 % de variación en las pulsaciones por minuto. Mientras que las personas que no hacen deporte habitualmente es de tan solo un 0.21 %.

## Pregunta 4

Calcula el porcentaje medio de variación del pulso por minuto entre antes y después de hacer ejercicio para los estudiantes que hacen ejercicio habitualmente y compara el valor de los hombres con el de las mujeres. ¿Observas mucha diferencia?

```python
df[df["hace.deporte"]=="si"].groupby("genero")["variacion_pulso"].mean()
```

La diferencia más significativa es en las personas que sí realizan deporte habitualmente. Los hombres tienen una variación media del 41.57 %, mientras que la de las mujeres es del 19.18 %, más de la mitad que los hombres.

## Pregunta 5

Calcula el porcentaje medio de variación del pulso por minuto entre antes y después de hacer ejercicio para los estudiantes que no hacen ejercicio habitualmente y compara el valor de los fumadores con los no fumadores. ¿Observas mucha diferencia?

```python
df[df["hace.deporte"]=="no"].groupby(["hace.deporte", "fuma"])["variacion_pulso"].mean()
```

La variación media de las pulsaciones por minuto de las personas que no hacen ejercicio habitualmente es baja, aunque en fumadores (0.91 %) es significativamente superior a los no fumadores (-0.0065 %).

## Pregunta 6

Calcula el porcentaje medio de variación del pulso por minuto entre antes y después de hacer ejercicio de todos los estudiantes según el tipo de actividad física que realizan. ¿Observas alguna diferencia?

```python
df.groupby("tipo.actividad")["variacion_pulso"].mean()
```

La actividad moderada es la que más variación de pulso representa con un 14.02 %. En tipos de actividades nulas el porcentaje de media de pulsaciones es ligeramente inferior con un 12.5 %. En actividades suaves e intensas las variaciones de pulso no son representativas.
