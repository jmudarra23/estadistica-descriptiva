# Reglas de Agrupación de Datos Numéricos

```python
import numpy as np
import pandas as pd
```

```python
def raiz(data):
    n = len(data)
    k1 = int(np.ceil(np.sqrt(n)))    
    return k1

def sturges(data):
    n = len(data)
    k2 = int(np.ceil(1 + np.log2(n)))
    return k2

def scott(data):
    n = len(data)
    As = 3.5*np.std(data)*n**(-1/3)
    k3 = int(np.ceil((max(data)-min(data))/As))
    return k3

def freedman_diaconis(data):
    n = len(data)
    irq = np.percentile(data, 75) - np.percentile(data, 25)
    Afd = 2 * irq * np.power(n, -1/3)
    k4 = int(np.ceil((max(data)-min(data) / Afd)))
    return k4
```

Cargamos el dataset:

```python
crabs = pd.read_table("../../data/datacrab.txt", sep = " ")
cw = np.array(crabs["width"])
```

## Pregunta 1

Da el algoritmo para reproducir el proceso de generación de clases y sus marcas donde el número de clases ha sido obtenido con la regla de la Scott en `Python`.

```python
# Cálculo de la amplitud
A = round((max(cw)-min(cw))/scott(cw), 1)

print("La amplitud es {}".format(A))

# Extremos de los intervalos
L = np.zeros(scott(cw) + 1)
L[0] = min(cw) - 1 / 2 * 0.1
for i in range(1, scott(cw) + 1):
    L[i] = L[0] + i * A

print("Los intervalos son: {}".format(L))
# Marcas de clase

X = np.zeros(scott(cw))
X[0] = (L[0] + L[1])/2
for i in range(1, scott(cw)):
    X[i] = (L[i] + L[i+1])/2
    
print("Las marcas son: {}".format(X))
```
