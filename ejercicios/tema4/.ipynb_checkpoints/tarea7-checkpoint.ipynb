# Gráficos con Python

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Pregunta 1

A partir del siguiente código, intenta obtener el gráfico del dibujo.

```python
n = 256
X = np.linspace(-np.pi, np.pi, n, endpoint = True)
Y = np.sin(2*X)

plt.plot(X, Y+1, color="blue", alpha=1.00)
plt.fill_between(X, 1, Y+1, color="blue", alpha=0.25)
plt.plot(X, Y-1, color="blue", alpha=1.00)
plt.fill_between(X, -1, Y-1, (Y-1) < -1, color="red", alpha=0.25)
plt.fill_between(X, -1, Y-1, (Y-1) > -1, color="blue", alpha=0.25)

plt.xlim(-np.pi, np.pi)
plt.ylim(-2.5, 2.5)
plt.xticks([], []) # eliminar los valores de los ticks del eje X
plt.yticks([], []) # eliminar los valores de los ticks del eje Y

plt.show()
```

Pista: investiga el parámetro `fill_between` de `matplotlib`.

# Pregunta 2

A partir del siguiente código, intenta obtener el gráfico del dibujo.

```python
n = 1024
X = np.random.normal(0, 1, n)
Y = np.random.normal(0, 1, n)
plt.scatter(X, Y, c=np.arctan2(Y, X), alpha=0.5, cmap="jet") # este es el cmap más parecido que he encontrado
plt.axis([-1.5, 1.5, -1.5, 1.5])

plt.xticks([], []) # eliminar los valores de los ticks del eje X
plt.yticks([], []) # eliminar los valores de los ticks del eje Y

plt.show()
```

Pista: el color viene dado por el ángulo que forman x e y con el eje horizontal.

# Pregunta 3

A partir del siguiente código, intenta obtener el gráfico del dibujo.

```python
n = 12
X = np.arange(n)
Y1 = (1 - X / float(n)) * np.random.uniform(0.5, 1.0, n)
Y2 = (1 - X / float(n)) * np.random.uniform(0.5, 1.0, n)
plt.bar(X, Y1, facecolor = "#9999ff", edgecolor = "white")
plt.bar(X, -Y2, facecolor = "#ff9999", edgecolor = "white")

for x, y in zip(X, Y1):
    plt.text(x, y + 0.05, "%.2f" % y, ha = "center", va = "bottom") # eliminado desplazamiento eje X del texto

    
for x, y in zip(X, Y2):
    plt.text(x, - y - 0.15, "%.2f" % y, ha = "center", va = "bottom")

plt.ylim(-1.25, 1.25)
    
plt.xticks([], []) # eliminar los valores de los ticks del eje X
plt.yticks([], []) # eliminar los valores de los ticks del eje Y

plt.show()
```

Pista: cuidado con la alineación del texto en el centro de las barras.

# Pregunta 4

A partir del siguiente código, intenta obtener el gráfico del dibujo.

```python
Z = np.ones(20) # generar array de 20 números con valores a 1
Z[-1] = 1.5 # último quesito tendrá un tamaño ligeramente más grande que el resto

colors = np.linspace(0, 1, 20) # 20 colores entre el 0 y el 1

plt.pie(Z, explode=Z*0.07, # separación entre quesitos
        colors=[str(x) for x in colors], # añadir lista colores en formato string
        wedgeprops={"edgecolor": "k",'linewidth': 0.5}) # borde negro y fino de los quesitos

plt.xticks([], []) # eliminar los valores de los ticks del eje X
plt.yticks([], []) # eliminar los valores de los ticks del eje Y

plt.show()
```

Pista: tendrás que modificar la Z.
